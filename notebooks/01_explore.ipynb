## Marco Geoestadistico

```python
import geopandas as gpd
from pathlib import Path

carpeta = Path("/Users/lucasgarcia/Desktop/Dataton/demanda_servicios_salud_cdmx/data/raw/09_ciudaddemexico/conjunto_de_datos")

for shp in carpeta.glob("*.shp"):
    try:
        gdf = gpd.read_file(shp, rows=1)  # lee solo 1 fila, rápido
        print(shp.name, "→", gdf.columns.tolist())
    except Exception as e:
        print(shp.name, "→ ERROR:", e)
```

```python
ageb_geom = gpd.read_file(carpeta / "09a.shp")

print(ageb_geom.shape)          # CDMX tiene ~8,000-9,000 AGEBs urbanas; si el número es muy distinto, algo no cuadra
print(ageb_geom.geom_type.unique())  # debe decir "Polygon" o "MultiPolygon"
ageb_geom.plot()                 # visualmente debe verse como la silueta de CDMX con sus zonas internas
print(ageb_geom.crs)              # revisa antes de reproyectar
```

```python
ageb_geom = gpd.read_file(carpeta / "09a.shp").to_crs(4326)
ageb_rural = gpd.read_file(carpeta / "09ar.shp").to_crs(4326)

ageb_geom_completo = pd.concat([ageb_geom, ageb_rural], ignore_index=True)
ageb_geom_completo = gpd.GeoDataFrame(ageb_geom_completo, geometry="geometry", crs="EPSG:4326")
```

```python
ageb_geom_completo.plot()
```
